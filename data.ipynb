```python
import pandas as pd
import numpy as np
```

```python
# Creamos datos para 50 estudiantes
data = {
    'ID': range(1, 51),
    'Carrera': np.random.choice(['Ingeniería', 'Psicología', 'Administración', 'Derecho'], 50),
    'Asistencia': np.random.randint(60, 100, 50),
    'Nota_Promedio': np.round(np.random.uniform(1.0, 5.0, 50), 1),
    'Horas_Estudio': np.random.randint(5, 25, 50)
}
```

```python
df = pd.DataFrame(data)
```

```python
# Creamos una lógica simple para la columna "Resultado"
df['Resultado'] = df.apply(lambda x: 'En Riesgo' if x['Nota_Promedio'] < 3.0 or x['Asistencia'] < 75 else 'Aprobado', axis=1)
```

```python
# Guardamos el archivo
df.to_csv('estudiantes_universidad.csv', index=False)
print("¡Archivo 'estudiantes_universidad.csv' creado con éxito!")
```
